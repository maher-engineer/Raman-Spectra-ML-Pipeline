```python
import pandas as pd
import numpy as np

# 1. تحديد مسار ملف الأهداف الرسمي في كاجل (train_terms.tsv)
file_path = "train_terms.tsv"

print("⏳ جاري تحميل وتشريح مصفوفات البيانات السحابية إحصائياً، يرجى الانتظار...")

try:
    # 2. قراءة الملف الأساسي من الـ input
    df = pd.read_csv(file_path, sep="\t")
    print("✅ تم تحميل الملف بنجاح!")

    # 3. فصل الفئات الثلاث عن بعضها كخطوة فرز حتمية
    bpo_raw = df[df['aspect'] == 'BPO'][['EntryID', 'term']].copy()
    cco_raw = df[df['aspect'] == 'CCO'][['EntryID', 'term']].copy()
    mfo_raw = df[df['aspect'] == 'MFO'][['EntryID', 'term']].copy()

    for raw_df in [bpo_raw, cco_raw, mfo_raw]:
        raw_df.columns = ['protein_id', 'go_term']


    # 4. حساب تكرار كل وظيفة جينية فريدة (Unique GO Terms) داخل كل فئة
    bpo_counts = bpo_raw['go_term'].value_counts().reset_index(name='count')
    cco_counts = cco_raw['go_term'].value_counts().reset_index(name='count')
    mfo_counts = mfo_raw['go_term'].value_counts().reset_index(name='count')
    bpo_final = pd.merge(bpo_raw, bpo_counts, on='go_term', how='left')
    cco_final = pd.merge(cco_raw, cco_counts, on='go_term', how='left')
    mfo_final = pd.merge(mfo_raw, mfo_counts, on='go_term', how='left')
```
