```python
import numpy as np
import pandas as pd
from Bio import SeqIO
import matplotlib.pyplot as plt
#fn = '/kaggle/input/cafa-5-protein-function-prediction/Test (Targets)/testsuperset.fasta'
#Train_Se
fn = 'train_sequences.fasta'
sequences = SeqIO.parse(fn, "fasta")
l = np.array([len(seq) for seq in sequences])
sequences = SeqIO.parse(fn, "fasta")
l2 = np.array([seq.id for seq in sequences])
print(pd.Series(l).describe(percentiles=[0.001, 0.05, 0.01, 0.1, 0.25, 0.5, 0.75, 0.9, 0.95, 0.99, 0.999]))
for k in [1000, 2000, 3000, 5000, 10_000, 15_000, 20_000, 30_000]:
    print('Len GE:', k, 'Count:', (l > k).sum())
set_prev = set()
for k in [30_000, 20_000, 10_000, 8000]:
    print('Len GE:', k, set(l2[l > k]) - set_prev)
    set_prev = set(l2[l > k])
plt.figure(figsize=(20, 3))
plt.hist(l, bins=100)
plt.title('train lengths', fontsize=20)
plt.show()
plt.figure(figsize=(20, 3))
plt.hist([t for t in l if t < 3000], bins=100)
plt.title('trai lengths', fontsize=20)
plt.show()
```
